# pygridmappr Quick Start

This notebook demonstrates how to allocate geographic points to a grid layout using `pygridmappr`.

```python
import pandas as pd
from pygridmappr import points_to_grid, visualize_allocation, generate_sample_points
```

## 1. Basic allocation

Four points at the corners of a square, mapped to a 2×2 grid.

```python
pts = pd.DataFrame({
    'area_name': ['SW', 'SE', 'NE', 'NW'],
    'x': [0, 100, 100, 0],
    'y': [0, 0, 100, 100]
})

result = points_to_grid(pts, n_row=2, n_col=2, compactness=0.5)
result[['area_name', 'row', 'col']]
```

```python
fig, axes = visualize_allocation(result, n_row=2, n_col=2, label_column='area_name')
```

## 2. Compactness parameter

Controls the trade-off between geographic fidelity (`0.5`) and grid density (`1.0`).

```python
from pygridmappr import compare_compactness

pts = generate_sample_points(n_points=20, pattern='random', seed=42)
fig, axes = compare_compactness(pts, n_row=5, n_col=5, compactness_values=[0.0, 0.5, 1.0])
```

## 3. Spacer cells

Block specific grid positions to create visual separation (e.g., island from mainland).

```python
pts = generate_sample_points(n_points=25, pattern='cluster', seed=123)
spacers = [(1, 1), (1, 2), (2, 1)]

result = points_to_grid(pts, n_row=6, n_col=6, compactness=0.5, spacers=spacers)
fig, axes = visualize_allocation(result, n_row=6, n_col=6, spacers=spacers)
```

## 4. Quality metrics

```python
from pygridmappr import compute_allocation_quality

quality = compute_allocation_quality(result, n_row=6, n_col=6)
for metric, value in quality.items():
    print(f"{metric}: {value:.3f}")
```
